```python
from confluent_kafka import Producer
import json
import time
import random
import uuid
from datetime import datetime
producer = Producer({'bootstrap.servers': 'kafka:9092'})
topic = "weather_s"
print(f"Connected to Kafka broker at kafka:9092 — sending data to topic '{topic}'")
def delivery_report(err, msg):
    if err is not None:
        print(f"Delivery failed: {err}")
LOCATIONS_LIST = [
    {"location_id": "location 1", "location_name": "Weather Station North", "latitude": 30.0801, "longitude": 31.2505},
    {"location_id": "location 2", "location_name": "City Center Station", "latitude": 30.0444, "longitude": 31.2357},
    {"location_id": "location 3", "location_name": "Airport Station", "latitude": 30.1219, "longitude": 31.4055},
    {"location_id": "location 4", "location_name": "Industrial Area Monitor", "latitude": 30.1500, "longitude": 31.1000},
]
LOCATION_MAP = {loc["location_id"]: loc for loc in LOCATIONS_LIST}
SENSORS = [
    {"device_id": "sensor 1", "location_id": LOCATIONS_LIST[0]["location_id"]}, 
    {"device_id": "sensor 2", "location_id": LOCATIONS_LIST[1]["location_id"]},
    {"device_id": "sensor 3", "location_id": LOCATIONS_LIST[2]["location_id"]},
    {"device_id": "sensor 4", "location_id": LOCATIONS_LIST[0]["location_id"]}, 
    {"device_id": "sensor 5", "location_id": LOCATIONS_LIST[3]["location_id"]},
]
SEASONS = ["Winter", "Spring", "Summer", "Autumn"]
DAY_PERIODS = ["Morning", "Afternoon", "Evening", "Night"]
WIND_DIRECTIONS = ["N", "NE", "E", "SE", "S", "SW", "W", "NW"]
while True:
    sensor = random.choice(SENSORS)
    device_id = sensor["device_id"]
    location_id = sensor["location_id"]
    location_info = LOCATION_MAP[location_id]
    location_name = location_info["location_name"]
    latitude = location_info["latitude"]
    longitude = location_info["longitude"]
    data = {
        "measurement_id": str(uuid.uuid4()),
        "device_id": device_id,
        "location_id": location_id,
        "location_name": location_name,
        "temperature": round(random.uniform(10.0, 60.0), 2),
        "humidity": random.randint(20, 90),
        "pressure": random.randint(980, 1050),
        "air_quality": random.randint(0, 100),
        "wind_speed": round(random.uniform(0, 20), 1),
        "wind_direction": random.choice(WIND_DIRECTIONS),
        "rain_level_mm": round(random.uniform(0, 15), 2),
        "visibility_km": round(random.uniform(2, 10), 1),
        "uv_index": random.randint(0, 11),
        "cloud_coverage_percent": random.randint(0, 100),
        "noise_level": random.randint(30, 100),       
        "battery_level": random.randint(10, 100),     
        "season": random.choice(SEASONS),
        "day_period": random.choice(DAY_PERIODS),
        "timestamp": datetime.utcnow().isoformat(),
        "latitude": latitude,
        "longitude": longitude
    }
    try:
        producer.produce(
            topic=topic,
            key=device_id.encode('utf-8'),
            value=json.dumps(data).encode('utf-8'),
            callback=delivery_report
        )
        producer.poll(0)
        print(f"Sent measurement from {location_name} (Device: {device_id[:8]}...) at {data['timestamp']}")
    except Exception as e:
        print(f"Error during produce operation: {e}")
    time.sleep(3)
```
